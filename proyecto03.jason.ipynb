```python
import yfinance as yf


tikers= ["AAPL", "NVDA", "MSFT"]
datos= yf.download(tikers, period="1y")

precios= datos["Close"]
```

```python
precios.head()
```

```python
precios.tail()
```

## Estadísticas simples

```python
precios.describe()
```

## Gráficos interactivos

```python
import plotly.express as px
```

```python
px.line(precios,
        title= "Cotización Histórica: Apple, Microsoft, NVIDIA (últimos 12 meses)",
        labels= {"value": "Precio (USD)", "Date": "Fechas"}, template="plotly_dark")
```

```python
px.bar(precios.mean(),
       title= "Precio promedio de Apple, Nvidia y Microsoft (últimos 12 meses)",
       labels= {"value": "Precio promedio (USD)", "Ticker": "Empresa"},
       color= precios.mean().index,
       template="plotly_dark"
       )
```

```python
precios.values[1]
```

```python
inversion= precios / precios.values[0] * 1000
```

```python
fig= px.line(inversion,
        title= "Si invertías $1000, ¿Cuánto tendrias hoy ",
        labels= {"value": "Valor de $1000", "Date": "Fecha", "variable": "Empresa"},
        template= "plotly_dark")
```

```python
fig.write_html("inversion de 1000 USD.html")
```
